```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#loading data assigning column names
col_names = ['ID', 'Topic', 'Sentiment', 'Text']
train_df = pd.read_csv(r"C:\Users\abnvz\Downloads\twitter_validation.csv", names=col_names)
val_df = pd.read_csv(r"C:\Users\abnvz\Downloads\twitter_training.csv", names=col_names)

#combining the datasets
df = pd.concat([train_df, val_df], ignore_index=True)

#droping rows with missng values 
df.dropna(subset=['Sentiment', 'Topic'], inplace=True)

#Sentiment Distribution
plt.figure(figsize=(8, 6))
sns.countplot(
    data=df, 
    x='Sentiment', 
    hue='Sentiment', # Added hue
    order=df['Sentiment'].value_counts().index, 
    palette='viridis',
    legend=False     # Added legend=False
)
plt.title('Overall Sentiment Distribution')
plt.xlabel('Sentiment')
plt.ylabel('Count')
plt.show()

#top 10 topics by Volume
top_topics = df['Topic'].value_counts().head(10).index
plt.figure(figsize=(12, 6))
sns.countplot(
    data=df[df['Topic'].isin(top_topics)], 
    y='Topic', 
    hue='Topic',     # Added hue
    order=top_topics, 
    palette='Set2',
    legend=False     # Added legend=False
)
plt.title('Top 10 Topics by Tweet Volume')
plt.xlabel('Count')
plt.ylabel('Topic')
plt.show()

#sentiment Distribution for Top Topics
# (This one already had 'hue' assigned properly)
plt.figure(figsize=(14, 8))
sns.countplot(
    data=df[df['Topic'].isin(top_topics)], 
    y='Topic', 
    hue='Sentiment', 
    order=top_topics, 
    palette='viridis'
)
plt.title('Sentiment Distribution for Top 10 Topics')
plt.xlabel('Count')
plt.ylabel('Topic')
plt.legend(title='Sentiment')
plt.show()

#prepare summary statistics
summary = {
    'total_tweets': len(df),
    'unique_topics': df['Topic'].nunique(),
    'sentiment_counts': df['Sentiment'].value_counts().to_dict()
}
print(summary)
```
